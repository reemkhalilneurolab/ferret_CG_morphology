# Manuscript figures and tables

Every figure and table below is produced by running the actual analysis
code in this repository against the cached pipeline outputs in
`figures/results/` (or, where noted, against the raw SWC/L-Measure data).
Nothing here is a static copy pasted in from the manuscript -- each cell
computes its own result and displays it. Figures 1 and 6 are the two
exceptions: both are external graphics (a schematic brain diagram adapted
from published sources, and a Canva-made summary infographic) that are not
generated by any code in this repository, so those two cells simply load
and display the published PNG.


```python
import os, sys, pickle
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
from IPython.display import display, Image as IPyImage

REPO = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
os.chdir(REPO)
sys.path.insert(0, str(REPO))

import config
config.PLOT_STEP_FUNCTION = False
config.PLOT_NEURON = False

from analysis.supporting._common import (
    load_matrices, load_classes, load_detection_weights,
    combine as common_combine, ward_clusters,
)

RUN = REPO / 'figures' / 'results'
PICKLES, PLOTS = RUN / 'pickles', RUN / 'plots'
MANUSCRIPT_FIGS = PLOTS / 'manuscript_figures'

plt.rcParams.update({'figure.dpi': 110, 'axes.spines.top': False,
                     'axes.spines.right': False})
TIER = {'V1-V2': '#1a80bb', 'PMLS-PLLS': '#ea801c'}

classes = load_classes()
matrices = load_matrices()
weights = load_detection_weights()
combined = common_combine(matrices, list(matrices), 'mean', weights)
stored = pd.read_pickle(PICKLES / 'combined_detection_weighted_matrix.pkl')
combined = combined.loc[stored.index, stored.columns]
labels = classes.loc[combined.index, 'class'].to_numpy()
print('repository:', REPO)
print('neurons:', len(combined), '| combined matrix matches cached pipeline output:',
      np.allclose(combined.values, stored.values))
```

## Figure 4 -- Representative neuron reconstructions

The five representative neurons are selected deterministically -- nearest the
centroid of each dendrogram cluster, of each anatomical class, and of the
mid-level subgroup assigned to the V1/V2 cluster -- in the principal-coordinate
space of the combined descriptor distance matrix. Selection logic matches
`analysis/supporting/make_manuscript_figures.py`; each selected neuron is then
rendered live from its SWC reconstruction.


```python
import neurotopo.utils as utl
from neurotopo.neuron_processor import process_neuron
import glob

squared = np.asarray(combined, dtype=float) ** 2
n = len(squared)
centring = np.eye(n) - np.ones((n, n)) / n
gram = -0.5 * centring @ squared @ centring
values, vectors = np.linalg.eigh(gram)
order = np.argsort(values)[::-1]
values, vectors = values[order], vectors[:, order]
descriptor_coords = vectors[:, :2] * np.sqrt(values[:2])

clusters = ward_clusters(combined, 2)
frame = pd.DataFrame({'neuron': combined.index, 'cls': labels, 'cluster': clusters,
                      'PC1': descriptor_coords[:, 0], 'PC2': descriptor_coords[:, 1]})


def nearest(subset, label):
    centroid = subset[['PC1', 'PC2']].mean().to_numpy()
    distance = np.linalg.norm(subset[['PC1', 'PC2']].to_numpy() - centroid, axis=1)
    row = subset.iloc[int(np.argmin(distance))]
    return {'selection': label, 'neuron': row['neuron'], 'class': row['cls'], 'cluster': int(row['cluster'])}

picks = []
for cluster in sorted(frame['cluster'].unique()):
    picks.append(nearest(frame[frame['cluster'] == cluster], f'cluster {cluster} centroid'))
for cls in ('V1-V2', 'PMLS-PLLS'):
    picks.append(nearest(frame[frame['cls'] == cls], f'{cls} class centroid'))
majority = frame.groupby('cluster')['cls'].agg(lambda s: s.value_counts().idxmax()).to_dict()
frame['crosses'] = frame.apply(lambda r: majority[r['cluster']] != r['cls'], axis=1)
for cls in ('V1-V2', 'PMLS-PLLS'):
    subset = frame[frame['crosses'] & (frame['cls'] == cls)]
    if len(subset):
        picks.append(nearest(subset, f'{cls} in opposite cluster'))

selection = pd.DataFrame(picks)
display(selection)

swc_files = {Path(p).stem.replace('.CNG', ''): p
            for p in glob.glob(str(REPO / 'data' / 'Briggs' / '**' / '*.swc'), recursive=True)}


def swc_path_for(name):
    for stem, path in swc_files.items():
        norm = stem.replace(' ', '_').replace('.', '_').replace('-', '_')
        if norm == name:
            return path
    return None

fig, axes = plt.subplots(1, len(selection), figsize=(4 * len(selection), 4.2))
for ax, (_, row) in zip(axes, selection.iterrows()):
    path = swc_path_for(row['neuron'])
    if path is None:
        ax.axis('off'); ax.set_title(f"{row['neuron']}\n(SWC not found)", fontsize=8)
        continue
    result = process_neuron(path)
    swc_df = result[6]
    tmp_dir = RUN / 'plots' / '_fig4_tmp'
    tmp_dir.mkdir(exist_ok=True)
    utl.plot_swc(swc_df, row['neuron'], row['class'], str(tmp_dir))
    rendered = tmp_dir / f"{row['class']}_{row['neuron']}.png"
    if rendered.exists():
        ax.imshow(mpimg.imread(rendered))
    ax.axis('off')
    ax.set_title(f"{row['selection']}\n{row['neuron']}", fontsize=8)
fig.suptitle('Representative CG neuron reconstructions')
fig.tight_layout(); plt.show()
```

## Figure 5 -- Per-neuron median descriptor values by cortical tier

One observation per neuron (the median of its step-function values, not the
value at a single radius), tested with a two-sided Mann-Whitney U test,
Benjamini-Hochberg corrected across the seven descriptors.


```python
with (RUN / 'all_descriptors.pkl').open('rb') as fh:
    descriptors = pickle.load(fh)

DESCRIPTOR_NAMES = ['Tortuosity', 'Branching_Pattern', 'Wiring', 'Flux', 'Leaf', 'Spread', 'Energy']

from scipy.stats import mannwhitneyu
from statsmodels.stats.multitest import multipletests

records = []
for name, per_neuron in descriptors.items():
    for neuron, step_function in per_neuron.items():
        vals = [v for _, v in step_function if v != 0]
        if vals:
            records.append({'descriptor': name, 'neuron': neuron,
                            'class': classes.loc[neuron, 'class'], 'median': np.median(vals)})
medians = pd.DataFrame(records)

stat_rows = []
for name in DESCRIPTOR_NAMES:
    sub = medians[medians['descriptor'] == name]
    g1 = sub.loc[sub['class'] == 'V1-V2', 'median'].values
    g2 = sub.loc[sub['class'] == 'PMLS-PLLS', 'median'].values
    stat, p = mannwhitneyu(g1, g2, alternative='two-sided')
    rank_biserial = 1 - (2 * stat) / (len(g1) * len(g2))
    stat_rows.append({'descriptor': name, 'V1/V2 median': np.median(g1),
                      'PMLS/PLLS/21a median': np.median(g2), 'p_raw': p, 'r': rank_biserial})
stat_df = pd.DataFrame(stat_rows)
stat_df['p_adj'] = multipletests(stat_df['p_raw'], method='fdr_bh')[1]
p_lookup = dict(zip(stat_df['descriptor'], stat_df['p_adj']))

fig, axes = plt.subplots(2, 4, figsize=(15, 6.5))
for ax, name in zip(axes.ravel(), DESCRIPTOR_NAMES):
    sub = medians[medians['descriptor'] == name]
    groups = [sub.loc[sub['class'] == cls, 'median'].values for cls in TIER]
    bp = ax.boxplot(groups, labels=list(TIER), patch_artist=True, widths=0.55, showfliers=False)
    for patch, colour in zip(bp['boxes'], TIER.values()):
        patch.set_facecolor(colour); patch.set_alpha(0.6)
    for med_line in bp['medians']:
        med_line.set_color('k')
    p = p_lookup[name]
    p_str = f'p={p:.2e}' if p < 0.001 else (f'p={p:.3f}' if p < 0.05 else f'p={p:.2f} (n.s.)')
    ax.set_title(f'{name}\n{p_str}', fontsize=10)
    ax.tick_params(axis='x', labelsize=8)
axes.ravel()[-1].axis('off')
fig.suptitle('Per-neuron median descriptor value by tier (BH-corrected Mann-Whitney)', fontsize=11)
fig.tight_layout(); plt.show()
```

## Table -- Standard morphometric variables by cortical tier (Table 3)

Nine of the forty L-Measure variables, measured on the same 117
reconstructions, compared between tiers by Mann-Whitney with
Benjamini-Hochberg correction across these nine variables.


```python
raw = pd.read_csv(REPO / 'data' / 'Briggs' / 'lmeasure_data.csv')
raw.columns = [c.strip() for c in raw.columns]
raw['neuron_name'] = (raw['Filename'].astype(str)
                      .str.replace('.swc', '', regex=False)
                      .str.replace('.CNG', '', regex=False)
                      .str.replace(' ', '_', regex=False)
                      .str.replace('.', '_', regex=False)
                      .str.replace('-', '_', regex=False))
raw = raw[raw['neuron_name'].isin(classes.index)].drop_duplicates('neuron_name').set_index('neuron_name')
raw['class'] = classes.loc[raw.index, 'class']

NINE_VARS = {'N_stems': 'Primary dendrites', 'EucDistance': 'Distance from soma (um)',
            'Height': 'Height (um)', 'N_branch': 'Branches', 'N_tips': 'Terminals',
            'Branch_pathlength': 'Mean branch length (um)', 'Width': 'Field width (um)',
            'Depth': 'Depth (um)', 'Length': 'Total dendritic length (um)'}

rows = []
for col, label in NINE_VARS.items():
    g1 = raw.loc[raw['class'] == 'V1-V2', col].dropna().values
    g2 = raw.loc[raw['class'] == 'PMLS-PLLS', col].dropna().values
    _, p = mannwhitneyu(g1, g2, alternative='two-sided')
    rows.append({'Measurement': label, 'V1/V2': np.median(g1),
                'PMLS/PLLS/21a': np.median(g2), 'p_raw': p})
table3 = pd.DataFrame(rows)
table3['p_adj'] = multipletests(table3['p_raw'], method='fdr_bh')[1]
display(table3[['Measurement', 'V1/V2', 'PMLS/PLLS/21a', 'p_adj']]
        .style.format({'V1/V2': '{:.3g}', 'PMLS/PLLS/21a': '{:.3g}', 'p_adj': '{:.2e}'}))
```

## Table -- Clustering performance, both representations (Table 1)

Silhouette, Adjusted Rand Index, Normalized Mutual Information, Purity, and
Cophenetic correlation for the topological-descriptor and standard
morphometric representations, both evaluated on the same 117 neurons with
Ward linkage and a two-cluster cut.


```python
from analysis.data_analysis import DendrogramAnalysis
from analysis.lmeasure_analysis import to_per_object
from scipy.cluster.hierarchy import linkage, fcluster, cophenet
from scipy.spatial.distance import squareform, pdist
from sklearn.metrics import adjusted_rand_score, silhouette_score, normalized_mutual_info_score
from sklearn.preprocessing import StandardScaler

class_df_d = classes.loc[combined.index].rename_axis('neuron_name').reset_index()
class_df_d['class_color'] = class_df_d['class'].map(TIER)
dendro_d = DendrogramAnalysis(data=combined, neuron_class_df=class_df_d.copy(),
                              save_directory=str(PLOTS), descriptor_name='notebook_descriptor_combined',
                              is_distance_matrix=True, cutoff_dist=0)
assign_d = dendro_d.run_analysis()
ari_d = adjusted_rand_score(labels, assign_d['dendrogram_cluster'].to_numpy())

numeric = to_per_object(raw.select_dtypes(include=['number']))
import neurotopo.utils as utl
numeric = utl.drop_constant_and_low_variance_columns(numeric)
numeric = numeric.loc[[n for n in combined.index if n in numeric.index]]
scaled = pd.DataFrame(StandardScaler().fit_transform(numeric), index=numeric.index, columns=numeric.columns)
lm_distance = pd.DataFrame(squareform(pdist(scaled.values, metric='euclidean')),
                           index=numeric.index, columns=numeric.index)
class_df_m = classes.loc[lm_distance.index].rename_axis('neuron_name').reset_index()
class_df_m['class_color'] = class_df_m['class'].map(TIER)
dendro_m = DendrogramAnalysis(data=lm_distance, neuron_class_df=class_df_m.copy(),
                              save_directory=str(PLOTS), descriptor_name='notebook_morphometric',
                              is_distance_matrix=True, cutoff_dist=0)
assign_m = dendro_m.run_analysis()
lm_labels = classes.loc[lm_distance.index, 'class'].to_numpy()
ari_m = adjusted_rand_score(lm_labels, assign_m['dendrogram_cluster'].to_numpy())


def purity(cluster_labels, truth):
    table = pd.crosstab(cluster_labels, truth)
    return table.max(axis=1).sum() / table.values.sum()

Z_d = linkage(squareform(combined.values, checks=False), method='ward')
Z_m = linkage(squareform(lm_distance.values, checks=False), method='ward')

table1 = pd.DataFrame({
    'Morphometric': {
        'Silhouette': silhouette_score(lm_distance.values, lm_labels, metric='precomputed'),
        'ARI': ari_m,
        'NMI': normalized_mutual_info_score(lm_labels, assign_m['dendrogram_cluster'].to_numpy()),
        'Purity': purity(assign_m['dendrogram_cluster'], pd.Series(lm_labels, index=lm_distance.index)),
        'Cophenetic correlation': cophenet(Z_m, squareform(lm_distance.values, checks=False))[0],
    },
    'Topological descriptors': {
        'Silhouette': silhouette_score(combined.values, labels, metric='precomputed'),
        'ARI': ari_d,
        'NMI': normalized_mutual_info_score(labels, assign_d['dendrogram_cluster'].to_numpy()),
        'Purity': purity(assign_d['dendrogram_cluster'], pd.Series(labels, index=combined.index)),
        'Cophenetic correlation': cophenet(Z_d, squareform(combined.values, checks=False))[0],
    },
})
display(table1.style.format('{:.3f}'))
print('Resampling-stability row (400 replicates, 80% subsampling) is fixed-seed but slow;')
print('see analysis/supporting/run_representation_stability_vs_morphometric.py for a live run.')
```

## Table -- Cluster composition by representation (TableComp)

Composition of the two clusters under each representation. Both are
obtained by Ward linkage with a two-cluster cut on the same 117 neurons.


```python
def composition(assign_df, label_series):
    merged = assign_df.set_index('neuron_name')[['dendrogram_cluster']].copy()
    merged['class'] = label_series
    majority = merged.groupby('dendrogram_cluster')['class'].agg(lambda s: s.value_counts().idxmax())
    cross = pd.crosstab(merged['dendrogram_cluster'], merged['class'])
    cross.index = ['V1/V2-dominated' if majority[i] == 'V1-V2' else 'extrastriate-dominated'
                  for i in cross.index]
    return cross.rename(columns={'V1-V2': 'V1/V2', 'PMLS-PLLS': 'PMLS/PLLS/21a'})

comp_d = composition(assign_d, pd.Series(labels, index=combined.index)).assign(Representation='Topological descriptors')
comp_m = composition(assign_m, pd.Series(lm_labels, index=lm_distance.index)).assign(Representation='Standard morphometric')
table_comp = pd.concat([comp_d, comp_m]).reset_index().rename(columns={'index': 'Cluster'})
table_comp = table_comp[['Representation', 'Cluster', 'PMLS/PLLS/21a', 'V1/V2']]
display(table_comp)
```
